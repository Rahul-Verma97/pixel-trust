# pixel-trust v2: trust maps vs. published uncertainty signals

Preliminary experiments for the proposal *Knowing Which Generated Pixels to Trust* (Rahul Verma).

**Setting.** KITTI Stereo 2015 (200 real pairs). The left image is warped into the right camera with a Depth Anything V2 depth prior aligned to sparse LiDAR (10% of the ground-truth points); Stable Diffusion inpainting fills the pixels the warp cannot supply (K = 4 samples). The real right image is the ground truth.

**v2 adds published baselines**, each calibrated and scored identically: multi-sample variance (Conffusion / ensembles), trajectory variance (Aithal et al., NeurIPS 2024), semantic entropy (Bose et al., ICCV 2025, with SegFormer-Cityscapes), and cross-view geometric confidence from VGGT (CVPR 2025), the verification idea of OracleGS / ConFixGS.

**Experiments.** Thrust 1: trust signals, sparsification/AUSE, conformal risk control on held-out images. Thrust 2: off-road-style depth corruption, depth-reliability signal, regeneration. Thrust 3: trust quality vs number of samples.

Settings: Accelerator **GPU T4 x2**, Internet **on**.

```python
# Smoke test: 2 images on GPU 0, to confirm every model and baseline loads before the full run
import glob, numpy as np
!CUDA_VISIBLE_DEVICES=0 python run_generate.py --root {root} --out /kaggle/working/smoke --limit 2 --k 4 --steps 25 2>&1 | tail -20
for f in sorted(glob.glob('/kaggle/working/smoke/feats/*.npz'))[:2]:
    z = np.load(f); print(f.split('/')[-1], sorted(k for k in z.files if k.startswith('f_')))
```

```python
# Stage 1: generate views + trust features, one process per GPU
import time
OUT = "/kaggle/working/out"
import torch
ngpu = max(1, torch.cuda.device_count())
procs = []
for g in range(ngpu):
    env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(g))
    log = open(f"/kaggle/working/gen_shard{g}.log", "w")
    procs.append(subprocess.Popen([sys.executable, "run_generate.py", "--root", root, "--out", OUT,
                                   "--shard", str(g), "--nshards", str(ngpu), "--k", "4", "--steps", "25"],
                                  stdout=log, stderr=subprocess.STDOUT, env=env))
t0 = time.time()
while any(p.poll() is None for p in procs):
    time.sleep(120)
    tails = [open(f"/kaggle/working/gen_shard{g}.log").read().strip().splitlines()[-1:] for g in range(ngpu)]
    print(f"{(time.time()-t0)/60:5.1f} min", tails, flush=True)
print("exit codes:", [p.returncode for p in procs])
for g in range(ngpu):
    print(open(f"/kaggle/working/gen_shard{g}.log").read()[-3000:])
```
